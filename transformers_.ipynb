```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="gpt2"
)
```

```python
prompt = "Deep learning will change the world because"

result = generator(
    prompt,
    max_length=50
)

print(result[0]["generated_text"])
```

```python
result = generator(
    prompt,
    max_length=80,
    temperature=0.8,
    top_k=50,
    top_p=0.95,
    num_return_sequences=3
)

for i, seq in enumerate(result):
    print(f"\n--- Génération {i+1} ---")
    print(seq["generated_text"])
```

```python
prompts = [
    "Artificial intelligence in education will",
    "In the future, humans and machines",
    "Climate change can be solved if"
]

for prompt in prompts:
    print("\nPrompt :", prompt)
    output = generator(prompt, max_length=60)
    print(output[0]["generated_text"])
```
