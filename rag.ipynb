```python
#RAG (Retrieval-Augmented Generation)

documents = [
    "Deep learning uses neural networks",
    "Transformers are attention-based models",
    "CNNs are good for images",
    "Transformers are widely used in NLP tasks"
]

query = "What are transformers?"

relevant_docs = [
    doc for doc in documents
    if "transformer" in doc.lower()
]

prompt = query + "\n\nContext:\n" + " ".join(relevant_docs)

def fake_llm(prompt: str) -> str:
    if "attention-based" in prompt.lower():
        return (
            "Transformers are neural network models based on attention mechanisms. "
            "They are commonly used in NLP tasks such as translation and text generation."
        )
    return "I don't have enough information to answer."

response = fake_llm(prompt)
```

```python
print("=== USER QUERY ===")
print(query)

print("\n=== RETRIEVED DOCUMENTS ===")
for doc in relevant_docs:
    print("-", doc)

print("\n=== PROMPT SENT TO MODEL ===")
print(prompt)

print("\n=== GENERATED ANSWER ===")
print(response)
```
